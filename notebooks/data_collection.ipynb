## Concatinating sources

```python
df1
```

```python
df4
```

```python
# Check column names to spot the differences
print("DF1 Columns:", df1.columns)
print("DF4 Columns:", df4.columns)

df1.rename(columns={
    'poem_text': 'البيت',      # Example: Change 'poem_text' to 'البيت'
    'poet_name': 'الشاعر',    # Example: Change 'poet_name' to 'الشاعر'
    'poem_style': 'العصر'            # Example: Change 'era' to 'العصر'
}, inplace=True)
```

```python

df = pd.concat([df1, df4], ignore_index=True)
```

```python
df
```

```python
df.isna().sum()
```

```python
df.drop("الديوان",axis=1,inplace=True)
```

```python
df
```

```python
# Find poems ('البيت') that appear more than once
conflicts = df[df.duplicated(subset=['البيت'], keep=False)]

# Sort them so the conflicting rows appear next to each other
conflicts = conflicts.sort_values(by='البيت')
```

```python
conflicts.head(50)
```

```python
print(f"Number of conflicting entries found: {len(conflicts)}")
display(conflicts[['الشاعر', 'البيت']].head(10))
```

```python

df.drop_duplicates(subset=['البيت'], keep='first', inplace=True)
df.reset_index(drop=True, inplace=True)
print(f"Final number of unique poems: {len(df)}")
```

```python

df.drop("poem_title",axis=1,inplace=True)
```

```python
df.to_csv("poetry_generation_dataset.csv")
```
